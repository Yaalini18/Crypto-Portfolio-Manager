```python
def calci(p):
    temp=[]
    for i in range(1,len(p)):
        change=((p[i]-p[i-1])/p[i-1])*100
        temp.append(change)
    return temp
p=[100,150,200,250]
print(calci(p))
```

```python
def fun2(p,n=3):
    a=[]
    for i in range(len(p)-n+1):
        w=p[i:i+n]
        avg=sum(w)/n
        a.append(avg)
    return a
p=[10,20,30,40,50]
print(fun2(p,n=3))
```

```python
import numpy as np
def fun3(p):
    c=[]
    for i in range(1,len(p)):
        change=((p[i]-p[i-1])/p[i-1])*100
        c.append(change)
    v=np.std(c)
    return round(v,2)
p = [10,20,30,40,50]
print("volatility",fun3(p))
```
