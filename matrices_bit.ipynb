```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Bitcoin_historical_data_formatted.csv")
df = df.sort_values(by="timeOpen")
prices = df["close"].tolist()

def calci(p):
    temp = []
    for i in range(1, len(p)):
        change = ((p[i] - p[i-1]) / p[i-1]) * 100
        temp.append(change)
    return temp

def fun2(p, n=3):
    a = []
    for i in range(len(p) - n + 1):
        w = p[i:i+n]
        avg = sum(w)/n
        a.append(avg)
    return a

def fun3(p):
    c = []
    for i in range(1, len(p)):
        change = ((p[i] - p[i-1]) / p[i-1]) * 100
        c.append(change)
    v = np.std(c)
    return round(v, 2)

percent_changes = calci(prices)
moving_avg = fun2(prices, n=3)
volatility = fun3(prices)

result_df = pd.DataFrame({
    "Close Price": prices,
    "Percent Change (prev day)": [None] + percent_changes,
    "3-day Moving Average": moving_avg + [None, None]
})

print(result_df)
print("\nVolatility (std dev of % changes):", volatility)
```
